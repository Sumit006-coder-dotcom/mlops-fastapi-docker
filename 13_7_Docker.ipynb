## 13.7.1 – Introduction to Docker

#### 1. What is Docker?

Docker is an open-source platform used to build, package, and run applications inside isolated environments called containers.

It packages an application along with its dependencies, libraries, and configuration.

Example:

Suppose your ML application works on your laptop but fails on another computer because of different Python versions or missing libraries.

Docker solves this problem by packaging the application and its required environment into a container.

#### 2. Why Do We Need Docker in MLOps?

Without Docker:

- Different Python versions can cause compatibility issues.

- Missing dependencies can break applications.

- Environment configuration must be repeated.

- Deployment becomes difficult across systems.

With Docker:

- Consistent application environments.

- Dependency management.

- Easy deployment.

- Application isolation.

- Better scalability.

- Portability across environments.

#### 3. Important Docker Terminology

| Term | Meaning |
|---|---|
| Docker | Platform for containerization |
| Dockerfile | Instructions to build an image |
| Docker Image | Blueprint containing application and dependencies |
| Docker Container | Running instance of an image |
| Docker Hub | Registry for storing and sharing images |
| Docker Engine | Software responsible for building and running containers |
| Docker Volume | Persistent storage for container data |
| Docker Network | Allows containers to communicate |

#### 4. Docker Image vs Container

| Docker Image | Docker Container |
|---|---|
| Blueprint of application | Running instance of image |
| Static template | Executable environment |
| Used to create containers | Runs the application |
| Can create multiple containers | Each container is isolated |

#### 5. Docker vs Virtual Machine

| Docker Container | Virtual Machine |
|---|---|
| Shares host OS kernel | Runs its own guest OS |
| Lightweight | Generally heavier |
| Starts quickly | Usually takes longer to start |
| Uses fewer resources | Uses more resources |
| Application-level isolation | Full OS-level virtualization |



## 13.7.2 – Docker Images, Containers and Essential Commands.

#### 1. Docker Images

A Docker image is a read-only template containing application code, dependencies, libraries and configuration required to run an application.

Example:
docker pull python:3.11-slim

#### 2. Docker Containers
A container is a running instance of a Docker image.

Example:
docker run hello-world

#### 3. Important Docker Commands

| Command | Purpose |
|---|---|
| `docker --version` | Check Docker version |
| `docker info` | Display Docker system information |
| `docker images` | List downloaded images |
| `docker ps` | List running containers |
| `docker ps -a` | List all containers |
| `docker pull python` | Download Python image |
| `docker run hello-world` | Run a test container |
| `docker stop <id>` | Stop a container |
| `docker start <id>` | Start a stopped container |
| `docker rm <id>` | Remove a container |
| `docker rmi <image>` | Remove an image |
| `docker logs <id>` | View container logs |

## Docker Volumes and Data Persistence.

#### Docker Volumes

#### 1. Concept
A Docker container has its own filesystem. Data stored inside a container can be lost when the container is removed.

Docker Volumes provide persistent storage outside the container's writable layer.

#### 2. Types of Storage
| Storage | Description |
|---|---|
| Docker Volume | Managed by Docker; persists beyond container removal. |
| Bind Mount | Maps a specific host directory or file into a container. |
| tmpfs Mount | Stores data in host memory; temporary. |

## Docker Compose.

#### 1. What is Docker Compose?
Docker Compose is a tool used to define and manage multiple containers using a single YAML configuration file.

For example, an ML application may require:
- FastAPI application
- MLflow tracking server
- PostgreSQL database
Instead of starting each container separately, Docker Compose can manage them together.
2. Important Concepts

| Concept | Description |
|---|---|
| `services` | Defines application containers |
| `image` | Specifies the Docker image |
| `build` | Builds an image using a Dockerfile |
| `ports` | Maps host and container ports |
| `volumes` | Provides persistent storage |
| `environment` | Defines environment variables |
| `depends_on` | Defines service startup dependencies |

#### Docker vs Docker Compose

| Docker | Docker Compose |
|---|---|
| Manages individual containers | Manages multiple services |
| Uses `docker run` | Uses `docker compose up` |
| Configuration through commands | Configuration through YAML |
| Suitable for individual containers | Useful for multi-container applications |
